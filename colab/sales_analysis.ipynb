```python
import pandas as pd

df = pd.read_csv("SalesMate AI - Lead Dataset - Sheet1.csv")
df.head()
```

```python
# Count the total number of leads
total_leads = len(df)

print("Total Number of Leads:", total_leads)
```

```python
# Count HOT, WARM and COLD leads
category_counts = df["Lead Category"].value_counts()

print("Lead Category Distribution:")
print(category_counts)
```

```python
# Count the most requested/recommended services
service_counts = df["Recommended Service"].value_counts()

print("Most Requested Services:")
print(service_counts)
```

```python
# Calculate the average lead score
average_score = df["Lead Score"].mean()

print("Average Lead Score:", round(average_score, 2))
```

```python
# Calculate the conversion rate
converted_leads = (df["Conversion Status"] == "Converted").sum()
conversion_rate = (converted_leads / len(df)) * 100

print("Converted Leads:", converted_leads)
print("Conversion Rate:", round(conversion_rate, 2), "%")
```

```python
# Count enquiries from each industry
industry_counts = df["Industry"].value_counts()

print("Industry-wise Enquiries:")
print(industry_counts)
```

```python
import matplotlib.pyplot as plt

# Create a chart of lead categories
category_counts = df["Lead Category"].value_counts()

category_counts.plot(kind="bar")

plt.title("Lead Category Distribution")
plt.xlabel("Lead Category")
plt.ylabel("Number of Leads")
plt.xticks(rotation=0)
plt.show()
```

```python
import matplotlib.pyplot as plt

# Count the recommended services
service_counts = df["Recommended Service"].value_counts()

# Create the chart
service_counts.plot(kind="bar")

plt.title("Most Requested Services")
plt.xlabel("Service")
plt.ylabel("Number of Leads")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

# Count enquiries by industry
industry_counts = df["Industry"].value_counts()

# Create the chart
industry_counts.plot(kind="bar")

plt.title("Industry-wise Enquiries")
plt.xlabel("Industry")
plt.ylabel("Number of Leads")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

# Count each conversion status
conversion_counts = df["Conversion Status"].value_counts()

# Create the chart
conversion_counts.plot(kind="bar")

plt.title("Conversion Status")
plt.xlabel("Status")
plt.ylabel("Number of Leads")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```
